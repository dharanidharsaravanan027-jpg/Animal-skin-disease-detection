```python
# ============================================================
# ANIMAL SKIN DISEASE DETECTION
# FINAL ROBUST GOOGLE COLAB VERSION
#
# Classes:
#   1. Healthy
#   2. ringworm
#   3. Mange
#
# Input  : Animal skin image
# Output : Predicted class + confidence
# ============================================================

import os
import shutil
import zipfile
import random
import warnings

warnings.filterwarnings("ignore")

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from google.colab import files
from tensorflow.keras import layers
from tensorflow.keras import models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

from sklearn.metrics import (
    confusion_matrix,
    classification_report
)

import seaborn as sns


# ============================================================
# 1. BASIC SETTINGS
# ============================================================

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 10
SEED = 42

EXPECTED_CLASSES = [
    "Healthy",
    "ringworm", # Changed to lowercase 'ringworm' to match folder name
    "Mange"
]

VALID_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
)

tf.random.set_seed(SEED)
np.random.seed(SEED)
random.seed(SEED)

print("=" * 60)
print("ANIMAL SKIN DISEASE DETECTION")
print("=" * 60)

print("\nTensorFlow version:", tf.__version__)


# ============================================================
# 2. UPLOAD DATASET ZIP
# ============================================================

print("\n" + "=" * 60)
print("STEP 1 - UPLOAD DATASET ZIP")
print("=" * 60)

print("""
Your ZIP should contain:

Animal_Skin_Dataset/
    Healthy/
    Ringworm/
    Mange/
""")

uploaded = files.upload()

zip_files = [
    name for name in uploaded.keys()
    if name.lower().endswith(".zip")
]

if len(zip_files) == 0:
    raise ValueError(
        "\nERROR: No ZIP file found.\n"
        "Please upload your dataset as a .zip file."
    )

zip_file = zip_files[0]

print("\nDataset ZIP selected:", zip_file)


# ============================================================
# 3. EXTRACT DATASET
# ============================================================

print("\nExtracting dataset...")

EXTRACT_DIR = "/content/animal_dataset"

if os.path.exists(EXTRACT_DIR):
    shutil.rmtree(EXTRACT_DIR)

os.makedirs(EXTRACT_DIR)

try:

    with zipfile.ZipFile(zip_file, "r") as zip_ref:
        zip_ref.extractall(EXTRACT_DIR)

except zipfile.BadZipFile:

    raise ValueError(
        "ERROR: ZIP file is corrupted. "
        "Please create the ZIP again."
    )


print("Dataset extracted successfully.")


# ============================================================
# 4. FIND CLASS FOLDERS AUTOMATICALLY
# ============================================================

print("\nSearching for class folders...")

dataset_path = None

for root, dirs, file_names in os.walk(EXTRACT_DIR):

    folder_names_lower = {
        d.lower(): d for d in dirs
    }

    if all(
        class_name.lower() in folder_names_lower
        for class_name in EXPECTED_CLASSES
    ):

        dataset_path = root
        break


if dataset_path is None:

    print("\nERROR: Required folders not found.")

    print("\nRequired folders:")

    for c in EXPECTED_CLASSES:
        print(" -", c)

    print("\nPlease check your ZIP structure.")

    raise ValueError(
        "Healthy, Ringworm and Mange folders were not found."
    )


print("\nDataset folder found:")
print(dataset_path)


# ============================================================
# 5. CHECK DATASET
# ============================================================

print("\n" + "=" * 60)
print("STEP 2 - DATASET CHECK")
print("=" * 60)

class_directories = {}

for expected_class in EXPECTED_CLASSES:

    found_directory = None

    for directory_name in os.listdir(dataset_path):

        if directory_name.lower() == expected_class.lower():

            full_path = os.path.join(
                dataset_path,
                directory_name
            )

            if os.path.isdir(full_path):

                found_directory = full_path
                break

    if found_directory is None:

        raise ValueError(
            f"Missing class folder: {expected_class}"
        )

    class_directories[expected_class] = found_directory


total_images = 0
class_counts = {}

for class_name in EXPECTED_CLASSES:

    folder = class_directories[class_name]

    image_files = []

    for file_name in os.listdir(folder):

        if file_name.lower().endswith(VALID_EXTENSIONS):

            image_files.append(file_name)

    count = len(image_files)

    class_counts[class_name] = count

    total_images += count

    print(
        f"{class_name:12s} : {count} images"
    )


print("-" * 40)
print("Total images :", total_images)


# ============================================================
# 6. CHECK MINIMUM DATA
# ============================================================

minimum_images_per_class = 5

for class_name, count in class_counts.items():

    if count < minimum_images_per_class:

        raise ValueError(
            f"\nERROR: {class_name} contains only "
            f"{count} images.\n"
            f"Please add at least "
            f"{minimum_images_per_class} images."
        )


# ============================================================
# 7. CHECK CLASS BALANCE
# ============================================================

largest_class = max(class_counts.values())
smallest_class = min(class_counts.values())

if smallest_class < largest_class * 0.5:

    print("\nWARNING:")
    print(
        "Your dataset is somewhat imbalanced."
    )

    print(
        "Try to keep approximately similar "
        "numbers of images in all classes."
    )

else:

    print("\nDataset class balance looks acceptable.")


# ============================================================
# 8. LOAD DATASET
# ============================================================

print("\n" + "=" * 60)
print("STEP 3 - LOADING DATASET")
print("=" * 60)

try:

    train_ds = tf.keras.utils.image_dataset_from_directory(

        dataset_path,

        labels="inferred",

        label_mode="int",

        class_names=EXPECTED_CLASSES,

        validation_split=0.20,

        subset="training",

        seed=SEED,

        image_size=IMG_SIZE,

        batch_size=BATCH_SIZE,

        shuffle=True
    )


    val_ds = tf.keras.utils.image_dataset_from_directory(

        dataset_path,

        labels="inferred",

        label_mode="int",

        class_names=EXPECTED_CLASSES,

        validation_split=0.20,

        subset="validation",

        seed=SEED,

        image_size=IMG_SIZE,

        batch_size=BATCH_SIZE,

        shuffle=False
    )

except Exception as e:

    raise RuntimeError(
        "\nDataset loading failed.\n"
        "Check your images and folder structure.\n\n"
        f"Original error: {e}"
    )


class_names = train_ds.class_names

print("\nClasses:", class_names)


# ============================================================
# 9. PREFETCH
# ============================================================

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(
    buffer_size=AUTOTUNE
)

val_ds = val_ds.prefetch(
    buffer_size=AUTOTUNE
)


# ============================================================
# 10. DATA AUGMENTATION
# ============================================================

data_augmentation = tf.keras.Sequential([

    layers.RandomFlip(
        "horizontal"
    ),

    layers.RandomRotation(
        0.10
    ),

    layers.RandomZoom(
        0.10
    ),

    layers.RandomContrast(
        0.10
    )

])


# ============================================================
# 11. MOBILE NET V2
# ============================================================

print("\n" + "=" * 60)
print("STEP 4 - BUILDING AI MODEL")
print("=" * 60)

try:

    base_model = MobileNetV2(

        input_shape=(
            IMG_SIZE[0],
            IMG_SIZE[1],
            3
        ),

        include_top=False,

        weights="imagenet"
    )

except Exception as e:

    raise RuntimeError(
        "Could not load MobileNetV2. "
        f"Error: {e}"
    )


base_model.trainable = False


# ============================================================
# 12. BUILD MODEL
# ============================================================

inputs = layers.Input(
    shape=(
        IMG_SIZE[0],
        IMG_SIZE[1],
        3
    )
)


x = data_augmentation(inputs)

x = preprocess_input(x)

x = base_model(
    x,
    training=False
)

x = layers.GlobalAveragePooling2D()(x)

x = layers.Dense(
    128,
    activation="relu"
)(x)

x = layers.Dropout(
    0.40
)(x)

outputs = layers.Dense(
    len(class_names),
    activation="softmax"
)(x)


model = models.Model(
    inputs,
    outputs
)


# ============================================================
# 13. COMPILE
# ============================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0001
    ),

    loss="sparse_categorical_crossentropy",

    metrics=[
        "accuracy"
    ]
)


print("\nModel created successfully.")

model.summary()


# ============================================================
# 14. TRAINING
# ============================================================

print("\n" + "=" * 60)
print("STEP 5 - TRAINING")
print("=" * 60)

try:

    history = model.fit(

        train_ds,

        validation_data=val_ds,

        epochs=EPOCHS
    )

except Exception as e:

    raise RuntimeError(
        "\nTraining failed.\n"
        f"Error: {e}"
    )


print("\nTraining completed successfully.")


# ============================================================
# 15. FINAL ACCURACY
# ============================================================

train_accuracy = (
    history.history["accuracy"][-1]
)

val_accuracy = (
    history.history["val_accuracy"][-1]
)

print("\n" + "=" * 60)
print("TRAINING RESULT")
print("=" * 60)

print(
    f"Training Accuracy   : "
    f"{train_accuracy * 100:.2f}%"
)

print(
    f"Validation Accuracy : "
    f"{val_accuracy * 100:.2f}%"
)


# ============================================================
# 16. ACCURACY GRAPH
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title(
    "Training and Validation Accuracy"
)

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.legend()

plt.grid(True)

plt.show()


# ============================================================
# 17. LOSS GRAPH
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "Training and Validation Loss"
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.legend()

plt.grid(True)

plt.show()


# ============================================================
# 18. CONFUSION MATRIX
# ============================================================

print("\n" + "=" * 60)
print("STEP 6 - CONFUSION MATRIX")
print("=" * 60)

y_true = []
y_pred = []


for images, labels in val_ds:

    predictions = model.predict(
        images,
        verbose=0
    )

    predicted_labels = np.argmax(
        predictions,
        axis=1
    )

    y_true.extend(
        labels.numpy()
    )

    y_pred.extend(
        predicted_labels
    )


cm = confusion_matrix(
    y_true,
    y_pred,
    labels=list(
        range(len(class_names))
    )
)


plt.figure(
    figsize=(7, 6)
)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title(
    "Confusion Matrix"
)

plt.xlabel(
    "Predicted Label"
)

plt.ylabel(
    "Actual Label"
)

plt.show()


# ============================================================
# 19. CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(

        y_true,

        y_pred,

        labels=list(
            range(len(class_names))
        ),

        target_names=class_names,

        zero_division=0
    )
)


# ============================================================
# 20. SAVE MODEL
# ============================================================

MODEL_PATH = (
    "/content/"
    "animal_skin_disease_model.keras"
)

try:

    model.save(
        MODEL_PATH
    )

    print(
        "\nModel saved successfully:"
    )

    print(
        MODEL_PATH
    )

except Exception as e:

    print(
        "\nWARNING: Model could not be saved."
    )

    print(e)


# ============================================================
# 21. UPLOAD TEST IMAGE
# ============================================================

print("\n" + "=" * 60)
print("STEP 7 - UPLOAD TEST IMAGE")
print("=" * 60)

print(
    "Upload a NEW animal skin image."
)

test_upload = files.upload()


test_image = None

for filename in test_upload.keys():

    if filename.lower().endswith(
        VALID_EXTENSIONS
    ):

        test_image = filename
        break


if test_image is None:

    raise ValueError(
        "\nNo valid image uploaded.\n"
        "Please upload JPG, JPEG, PNG, "
        "BMP or WEBP."
    )


print(
    "\nTest image:",
    test_image
)


# ============================================================
# 22. LOAD TEST IMAGE
# ============================================================

try:

    image = tf.keras.utils.load_img(

        test_image,

        target_size=IMG_SIZE
    )

    image_array = (
        tf.keras.utils.img_to_array(
            image
        )
    )

    image_array = np.expand_dims(
        image_array,
        axis=0
    )

except Exception as e:

    raise RuntimeError(
        "Could not read test image.\n"
        f"Error: {e}"
    )


# ============================================================
# 23. PREDICTION
# ============================================================

try:

    prediction = model.predict(
        image_array,
        verbose=0
    )[0]

except Exception as e:

    raise RuntimeError(
        "Prediction failed.\n"
        f"Error: {e}"
    )


predicted_index = int(
    np.argmax(prediction)
)

predicted_class = (
    class_names[predicted_index]
)

confidence = (
    float(
        prediction[predicted_index]
    ) * 100
)


# ============================================================
# 24. FINAL OUTPUT
# ============================================================

print("\n" + "=" * 60)
print("FINAL PREDICTION")
print("=" * 60)

print(
    "Uploaded Image :",
    test_image
)

print(
    "Prediction     :",
    predicted_class
)

print(
    f"Confidence     : "
    f"{confidence:.2f}%"
)


if predicted_class == "Healthy":

    print(
        "Result         : "
        "No disease detected"
    )

elif predicted_class == "Ringworm":

    print(
        "Result         : "
        "Ringworm detected"
    )

elif predicted_class == "Mange":

    print(
        "Result         : "
        "Mange detected"
    )


print("=" * 60)


# ============================================================
# 25. SHOW IMAGE + RESULT
# ============================================================

plt.figure(
    figsize=(7, 7)
)

plt.imshow(
    image
)

plt.axis("off")

plt.title(
    f"Prediction: {predicted_class}\n"
    f"Confidence: {confidence:.2f}%"
)

plt.show()


# ============================================================
# 26. ALL CLASS PROBABILITIES
# ============================================================

print(
    "\nProbability of each class:"
)

for i, class_name in enumerate(
    class_names
):

    print(
        f"{class_name:12s} : "
        f"{prediction[i] * 100:.2f}%"
    )


```
